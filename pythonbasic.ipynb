```python
import pandas as pd
import numpy as np
exam_data = {
    "name": ["Ajay", "Vijay", "Vivek", "Ramesh", "Abhijit", "Amit"],
    "score": [12.5, 9, 16.5, np.nan, 9, 20],
    "attempts": [1, 3, 2, 3, 2, 3],
    "qualify": ["yes", "no", "yes", "no", "yes", "no"]
    }
labels = ["a", "b", "c", "d", "e", "f"]
df = pd.DataFrame(exam_data, index=labels)
print("Original DataFrame:")
print(df)
print("\n(a) Name and Score columns:")
print(df[["name", "score"]])
print("\n(b) Students with score between 10 and 15: ")
print(df[df["score"].between(10, 15)])
print("\n(c) Rwos where score is missing:")
print(df[df["score"].isna()])
df.loc["d", "score"] = 11.5
print("\n(d) DataFrame after changing scoe of row 'd':")
print(df)
total_attempts = df["attempts"].sum()
print("\n(e) Total number of examination attempts:")
print(total_attempts)
```
